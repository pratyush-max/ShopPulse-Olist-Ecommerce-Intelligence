```python
# Reusable function to save a DataFrame as a Bronze Delta table

def save_bronze_table(df, table_name):

    # Get the short table name
    short_name = table_name.split(".")[-1]

    # Check whether the table already exists
    existing = spark.sql(
        f"SHOW TABLES IN workspace.default LIKE '{short_name}'"
    ).count()

    if existing > 0:
        print(f"{table_name} already exists. Skipping creation.")
        return

    # Create a temporary view
    source_view = f"{short_name}_source"
    df.createOrReplaceTempView(source_view)

    # Create the managed Delta table
    spark.sql(f"""
        CREATE TABLE {table_name}
        USING DELTA
        AS SELECT * FROM {source_view}
    """)

    print(f"Successfully created: {table_name}")


# Save the Customers DataFrame
save_bronze_table(
    customers_df,
    "workspace.default.bronze_customers"
)
```

```python
# Reusable function to verify Bronze Delta tables

def verify_bronze_table(table_name, expected_count):

    # Load the saved Delta table
    df = spark.table(table_name)

    # Count the saved records
    actual_count = df.count()

    print("Table:", table_name)
    print("Expected records:", expected_count)
    print("Actual records:", actual_count)

    # Check whether record counts match
    if actual_count == expected_count:
        print("Verification: PASSED")
    else:
        print("Verification: FAILED")

    # Display the table schema
    df.printSchema()

    # Check Delta format and storage details
    display(
        spark.sql(f"DESCRIBE DETAIL {table_name}")
        .select("format", "name", "numFiles", "sizeInBytes")
    )


# Verify Customers
verify_bronze_table(
    "workspace.default.bronze_customers",
    99441
)
```

```python
from pyspark.sql import functions as F

# Load the original Orders Bronze table
orders_bronze_df = spark.table(
    "workspace.default.bronze_orders"
)

# 1. Check total records and unique order IDs
display(
    orders_bronze_df.agg(
        F.count("*").alias("total_records"),
        F.countDistinct("order_id").alias("unique_order_ids"),
        F.sum(
            F.when(F.col("order_id").isNull(), 1).otherwise(0)
        ).alias("null_order_ids")
    )
)

# 2. Check missing values in original columns
original_columns = [
    c for c in orders_bronze_df.columns
    if not c.startswith("_")
]

null_summary = orders_bronze_df.select([
    F.sum(
        F.when(F.col(c).isNull(), 1).otherwise(0)
    ).alias(c)
    for c in original_columns
])

display(null_summary)

# 3. Check the distribution of order statuses
display(
    orders_bronze_df
    .groupBy("order_status")
    .count()
    .orderBy("order_status")
)
```

```python
# Convert the existing null summary into a readable vertical table
null_counts_dict = null_summary.first().asDict()

null_counts_list = [
    (column, int(count or 0))
    for column, count in null_counts_dict.items()
]

null_counts_vertical_df = spark.createDataFrame(
    null_counts_list,
    ["column_name", "null_count"]
)

# Display columns with the most missing values first
display(
    null_counts_vertical_df
    .orderBy(F.desc("null_count"))
)
```

```python
# Analyze missing dates for each order status
orders_missing_by_status = (
    orders_bronze_df
    .groupBy("order_status")
    .agg(
        F.count("*").alias("total_orders"),

        F.sum(
            F.when(
                F.col("order_approved_at").isNull(), 1
            ).otherwise(0)
        ).alias("missing_approval_date"),

        F.sum(
            F.when(
                F.col("order_delivered_carrier_date").isNull(), 1
            ).otherwise(0)
        ).alias("missing_carrier_date"),

        F.sum(
            F.when(
                F.col("order_delivered_customer_date").isNull(), 1
            ).otherwise(0)
        ).alias("missing_customer_delivery_date")
    )
    .orderBy("order_status")
)

display(orders_missing_by_status)
```

```python
# Find delivered orders with one or more missing dates
delivered_anomalies_df = (
    orders_bronze_df
    .filter(
        (F.col("order_status") == "delivered") &
        (
            F.col("order_approved_at").isNull() |
            F.col("order_delivered_carrier_date").isNull() |
            F.col("order_delivered_customer_date").isNull()
        )
    )
    .select(
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    )
    .orderBy("order_id")
)

# Display the exceptional records
print("Delivered orders with missing dates:", delivered_anomalies_df.count())

display(delivered_anomalies_df)
```

```python
# Check whether order timestamps follow the expected sequence

orders_sequence_df = (
    orders_bronze_df

    # Check if approval happened before purchase
    .withColumn(
        "purchase_after_approval",
        F.col("order_approved_at").isNotNull() &
        (F.col("order_purchase_timestamp") > F.col("order_approved_at"))
    )

    # Check if carrier delivery happened before approval
    .withColumn(
        "approval_after_carrier",
        F.col("order_delivered_carrier_date").isNotNull() &
        F.col("order_approved_at").isNotNull() &
        (F.col("order_approved_at") > F.col("order_delivered_carrier_date"))
    )

    # Check if customer delivery happened before carrier delivery
    .withColumn(
        "carrier_after_customer",
        F.col("order_delivered_customer_date").isNotNull() &
        F.col("order_delivered_carrier_date").isNotNull() &
        (F.col("order_delivered_carrier_date") > F.col("order_delivered_customer_date"))
    )
)

# Count timestamp sequence violations
display(
    orders_sequence_df.agg(
        F.sum(F.col("purchase_after_approval").cast("int"))
            .alias("purchase_after_approval"),

        F.sum(F.col("approval_after_carrier").cast("int"))
            .alias("approval_after_carrier"),

        F.sum(F.col("carrier_after_customer").cast("int"))
            .alias("carrier_after_customer")
    )
)
```

```python
# Group timestamp sequence violations by order status
sequence_issues_by_status = (
    orders_sequence_df
    .groupBy("order_status")
    .agg(
        F.count("*").alias("total_orders"),

        F.sum(
            F.col("approval_after_carrier").cast("int")
        ).alias("approval_after_carrier_issues"),

        F.sum(
            F.col("carrier_after_customer").cast("int")
        ).alias("carrier_after_customer_issues")
    )
    .orderBy("order_status")
)

display(sequence_issues_by_status)
```

```python

from pyspark.sql import functions as F

# Step 1: Identify valid carrier timestamps
carrier_valid = (
    F.col("order_delivered_carrier_date").isNull()
    |
    (
        (
            F.col("order_delivered_carrier_date")
            >= F.col("order_purchase_timestamp")
        )
        &
        (
            F.col("order_approved_at").isNull()
            |
            (
                F.col("order_delivered_carrier_date")
                >= F.col("order_approved_at")
            )
        )
    )
)

# Step 2: Identify valid customer delivery timestamps
# Compare against the original carrier timestamp only
# when that carrier timestamp is valid.

customer_valid = (
    F.col("order_delivered_customer_date").isNull()
    |
    (
        (
            F.col("order_delivered_customer_date")
            >= F.col("order_purchase_timestamp")
        )
        &
        (
            F.col("order_approved_at").isNull()
            |
            (
                F.col("order_delivered_customer_date")
                >= F.col("order_approved_at")
            )
        )
        &
        (
            F.col("order_delivered_carrier_date").isNull()
            |
            (~F.col("_carrier_date_valid"))
            |
            (
                F.col("order_delivered_customer_date")
                >= F.col("order_delivered_carrier_date")
            )
        )
    )
)

# Step 3: Create temporary validity flags
orders_flagged_df = (
    bronze_orders_df
    .withColumn("_carrier_date_valid", carrier_valid)
    .withColumn("_customer_date_valid", customer_valid)
)

# Step 4: Clean and standardize columns
silver_orders_df = (
    orders_flagged_df

    # Standardize string columns
    .withColumn("order_id", F.trim(F.col("order_id")))
    .withColumn("customer_id", F.trim(F.col("customer_id")))
    .withColumn(
        "order_status",
        F.lower(F.trim(F.col("order_status")))
    )

    # Remove invalid carrier timestamps
    .withColumn(
        "order_delivered_carrier_date",
        F.when(
            F.col("_carrier_date_valid"),
            F.col("order_delivered_carrier_date")
        ).otherwise(F.lit(None).cast("timestamp"))
    )

    # Remove invalid customer delivery timestamps
    .withColumn(
        "order_delivered_customer_date",
        F.when(
            F.col("_customer_date_valid"),
            F.col("order_delivered_customer_date")
        ).otherwise(F.lit(None).cast("timestamp"))
    )

    # Keep only business columns
    .select(
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    )
)

print("Silver Orders transformation completed!")
print("Original records:", bronze_orders_df.count())
print("Transformed records:", silver_orders_df.count())

# Inspect transformed schema
silver_orders_df.printSchema()

# Preview cleaned records
display(silver_orders_df.limit(10))
```

```python

# Identify seller reference exceptions
closed_deals_audit_source = (
    bronze_closed_deals_df
    .withColumn(
        "seller_id",
        F.trim(F.col("seller_id"))
    )
    .join(
        seller_reference_df,
        on="seller_id",
        how="left"
    )
)

# Create one audit record per deal
closed_deals_quality_audit_df = (
    closed_deals_audit_source
    .select(
        "mql_id",
        "seller_id",
        "_ingested_at",
        "_source_file",

        F.col("business_segment").isNull()
        .alias("is_missing_business_segment"),

        F.col("lead_type").isNull()
        .alias("is_missing_lead_type"),

        F.col("lead_behaviour_profile").isNull()
        .alias("is_missing_lead_behaviour_profile"),

        F.col("has_company").isNull()
        .alias("is_missing_company_information"),

        F.col("has_gtin").isNull()
        .alias("is_missing_gtin_information"),

        F.col("average_stock").isNull()
        .alias("is_missing_average_stock"),

        F.col("business_type").isNull()
        .alias("is_missing_business_type"),

        F.col("declared_product_catalog_size").isNull()
        .alias("is_missing_catalog_size"),

        (F.col("declared_monthly_revenue") == 0)
        .alias("has_zero_declared_revenue"),

        F.col("seller_exists").isNull()
        .alias("seller_reference_missing")
    )
)

print(
    "Closed Deals Quality Audit Records:",
    closed_deals_quality_audit_df.count()
)

display(closed_deals_quality_audit_df.limit(20))
```
